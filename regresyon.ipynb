```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
raw_df = pd.read_csv(r'C:\Regresyon\raw.csv')
test_df = pd.read_csv(r'C:\Regresyon\test.csv')
print("RAW:")
print(raw_df.head())
print("\nTEST:")
print(test_df.head())
```

```python
print(raw_df.shape)
print(test_df.shape)
```

```python
train_df = pd.read_csv('raw.csv', delim_whitespace=True, header=None)
test_df = pd.read_csv('test.csv', delim_whitespace=True, header=None)

X_train = train_df.iloc[:, 1:]
y_train = train_df.iloc[:, 0]

X_test = test_df.iloc[:, 1:]
y_test = test_df.iloc[:, 0]

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

predictions = model.predict(X_test)
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print(f"MSE: {mse}")
print(f"R2: {r2}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred, alpha=0.3, color='blue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Gerçek Değerler')
plt.ylabel('Tahmin Edilen Değerler')
plt.title('Gerçek Değerler vs Tahmin Edilen Değerler')
plt.show()
```
